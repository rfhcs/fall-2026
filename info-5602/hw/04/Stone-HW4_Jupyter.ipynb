```python
#Setup Section 1: Run This Cell first before you attempt Questions 1 - 3

#Look at the variables in the DataFrame once they display.

import pandas as pd
from pandas import read_csv
import seaborn as sns

#Obtain dataset and display data frame:
college_data = pd.read_csv("https://vincentarelbundock.github.io/Rdatasets/csv/AER/CollegeDistance.csv", index_col = 0).dropna()
college_data.head()

#Note the variable names in the college_data DataFrame
```

```python
#Setup Section 1.1: Run this cell before attempting Questions 1

#The code below draws the kernel density estimate (KDE) of variable score in the college_data dataframe
#conditioned on the variable region:


sns.kdeplot(x=college_data.score, hue = college_data.region)
```

```python
#Question 1:

#Note that region is a categorical variable containing only two categories/groups.

#A friend argues that conditioning the kernel density plot of variable score on region suggests that observations
#from one of the groups has a perfectly normal probability distribution function/curve. Do you agree? Why or why not?



#Write your answer below:

#I disagree. Conditioning the kernel density plot on one variable to another just visualizes the probability density as a continuous curve.
# The curve that results from plotting score on region may or may not be normal.

```

```python
#Setup Section 1.2: Run this cell before attempting Question 2.

#The code below draws the boxplot of variable score in the college_data DataFrame conditioned on the variable region:

sns.boxplot(y=college_data.score, x = college_data.region)

```

```python
#QUESTION 2:

#In Setup Section 1.2 above the boxplot of score conditioned on region, a variable that has two groups/categories.
#Which of those groups appears to have observations with the larger interquartile range (IQR)? 

#Choose one the options below:

#Option 1: other
#Option 2: west


#Write your code/answer below:

q3_other = college_data[college_data['region'] == 'other'].score.quantile(0.75)
q1_other = college_data[college_data['region'] == 'other'].score.quantile(0.25)

print("IQR of 'other':", q3_other - q1_other)

q3_west = college_data[college_data['region'] == 'west'].score.quantile(0.75)
q1_west = college_data[college_data['region'] == 'west'].score.quantile(0.25)

print("IQR of 'west':", q3_west - q1_west)

#CHOSEN ANSWER: Option 1: other

#The interquartile range is the measure of the data of Q3 - Q1. From the code above, the IQR of 'other' is greater than that of 'west'.

```

```python
#Setup Section 1.3: Run this cell before attempting Question 3.

#The code below draws the boxplot of variable unemp in the college_data dataframe
#conditioned on as follows:

#1. Condition the plot first on region 
#2. Condition the plot on home next

sns.boxplot(y=college_data.unemp, x= college_data.region, hue = college_data.home)

```

```python
#QUESTION 3:

#Consider your chart in setup section 1.3 above.

#What appears to be visible about the size of the interquartile range of variable unemp when:

#region = other
#home = yes

#compared to the interquartile range of variable unemp when:

#region = west
#home = no


#Write your answer below:

q3_other_home = college_data[(college_data['region'] == 'other') & (college_data['home'] == 'yes')].unemp.quantile(0.75)
q1_other_home = college_data[(college_data['region'] == 'other') & (college_data['home'] == 'yes')].unemp.quantile(0.25)

print("IQR for region = other, home = yes: ", q3_other_home - q1_other_home)

q3_west_home = college_data[(college_data['region'] == 'west') & (college_data['home'] == 'no')].unemp.quantile(0.75)
q1_west_home = college_data[(college_data['region'] == 'west') & (college_data['home'] == 'no')].unemp.quantile(0.25)

print("IQR for region = west, home = no: ", q3_west_home - q1_west_home)


#When region = other and home = yes , the size of the interquartile range is larger compared to region = west and home = no.
```

```python
#Setup Section 4: Run this cell before you attempt Questions 8 - 10

#Loading Matplotlib and Seaborn

from matplotlib import pyplot as plt
import seaborn as sns

#Load the mpg dataset and assigning to dataframe cars
cars = sns.load_dataset('mpg').dropna()

#Set seaborn style to "whitegrid" - display gridlines on white background for plots
sns.set_style('whitegrid')

#extract odd number of cylinders in cars. Store the result in a DataFrame called cars_odd 
cars_odd = cars[cars.cylinders.isin([3,5])]

#Breakout or condition the displacement variable on the types (category) of cylinders.
#Use horizontal axis for the categorical axis (cylinders) and vertical axis for the main variable (displacement)
sns.violinplot(x=cars_odd.cylinders, y=cars_odd.displacement)

#Split data - to use split feature, you must have exactly two values for the hue argument/parameter
# Split argument/parameter allows you compare two categories side-by-side to draw additional insights
#Dive deeper to see the quartiles of data using the inner argument i.e. set inner = 'quartiles'

# Place legend in location 2

sns.violinplot(x='cylinders',y='displacement',hue='origin',
              data=cars_odd[cars_odd.origin.isin(['japan','europe'])],
              split=True,
              inner='quartiles')

plt.legend(loc=2)
```
